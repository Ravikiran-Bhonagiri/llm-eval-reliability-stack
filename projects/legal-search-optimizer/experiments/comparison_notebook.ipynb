# Legal RAG Optimization Experiment

This notebook compares 3 chunking strategies to find the Pareto Frontier between Retrieval Recall and Cost.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Results from Grid Search Script
df = pd.read_csv('final_results.csv')
print(df)

# Visualization
plt.figure(figsize=(10, 6))
plt.scatter(df['config'], df['recall'], s=100, c='blue', label='Recall')
plt.scatter(df['config'], df['precision'], s=100, c='red', marker='x', label='Precision')
plt.title('Chunking Strategy Performance')
plt.ylabel('Score (0-1)')
plt.legend()
plt.grid(True)
plt.show()
```
